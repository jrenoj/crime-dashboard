```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/Users/jrenoj/Desktop/crime-dashboard/data/processed/processed.csv')
pd.set_option('display.max_columns', None)
df.head()
```

```python
df.groupby('Year').size().plot(kind='bar', figsize=(10, 6))
```

```python
df.groupby('AREA NAME').size().sort_values(ascending=False).plot(kind='bar', figsize=(10, 6))
```

```python
df.groupby("Month").size().sort_values(ascending=False).plot(kind='bar', figsize=(10, 6))
```

```python
df.groupby("Day").size().sort_values(ascending=False).plot(kind='bar', figsize=(10, 6))
```

```python
df.groupby("Vict Sex").size().sort_values(ascending=False).plot(kind='bar', figsize=(10, 6))
```

```python
df["Vict Age"].plot(kind='hist', bins=20,figsize=(10, 6), title='Distribution of Victim Age', xlabel='Age', ylabel='Frequency', color='skyblue', edgecolor='black')  
```

```python
df["Vict Age"] = df["Vict Age"].replace(0, np.nan)
```

```python
df["Vict Age"].plot(kind='hist', bins=40,figsize=(10, 6), title='Distribution of Victim Age', xlabel='Age', ylabel='Frequency', color='skyblue', edgecolor='black')  
```

```python
df["Time"].str.split(':').str[0].astype(int).plot(kind='hist', bins=24, figsize=(10, 6), title='Distribution of Crime Time', xlabel='Hour of the Day', ylabel='Frequency', color='skyblue', edgecolor='black')
```

```python
df["Crm Cd Desc"].value_counts().head(10).plot(kind='barh', figsize=(10, 6), title='Top 10 Crime Descriptions', xlabel='Crime Description', ylabel='Frequency', color='skyblue', edgecolor='black')
```
