```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/Users/jrenoj/Desktop/crime-dashboard/data/raw/export.csv')
pd.set_option('display.max_columns', None)
df.head()

```

```python
new = df[["DR_NO", "DATE OCC", "TIME OCC", "AREA NAME", "Crm Cd Desc", "Vict Age", "Vict Sex", "Vict Descent", "LAT", "LON"]]
new.head()
```

```python
new["Year"] = pd.to_datetime(new["DATE OCC"]).dt.year
new["Month"] = pd.to_datetime(new["DATE OCC"]).dt.month_name()
new["Day"] = pd.to_datetime(new["DATE OCC"]).dt.day_name()
new["Time"] = (
    pd.to_datetime(new["TIME OCC"].astype(str).str.zfill(4), format="%H%M")
    .dt.hour.astype(str)
    + ":00"
)

new.head()
```

```python
df = new.drop(columns=[new.columns[2]], axis=1)
df.to_csv('/Users/jrenoj/Desktop/crime-dashboard/data/processed/processed.csv', index=False)
```
